```python
import re
import numpy as np
D = 16
re = re.compile(r'[0-9.]+')
a,av,ar = np.loadtxt(f'{D}{D}{D}.csv','7B,f,B',delimiter=',',skiprows=1,usecols=(6,7,8,9,10,11,12,16,19),converters=lambda x:next(re.finditer(x)).group(),unpack=True,encoding='utf-8-sig')

N = np.arange(len(a),dtype='H')
m = (a[N,a[:,3]]==1)&(a[N,a[:,4]]==1)
av = np.concatenate([av,av[m]])
ar = np.concatenate([ar,ar[m]])
t = a[m]
t[:,3:5] = t[:,4:2:-1]
a = np.concatenate((a,t))

m = (a[:,:3]==1).all(1)
av = np.concatenate([av]+[av[m]]*2)
ar = np.concatenate([ar]+[ar[m]]*2)
t = a[m]
t[:,3] = 3 - t[:,3:5].sum(1)
m = a[m]
m[:,4] = 3 - t[:,3:5].sum(1)
a = np.concatenate((a,t,m))

N = np.arange(len(a),dtype='H')
m = (a[N,a[:,6]]==1)&(a[N,3-a[:,5:].sum(1)]==D)
av = np.concatenate([av,av[m]])
ar = np.concatenate([ar,ar[m]])
t = a[m]
t[:,6] = 3-t[:,5:].sum(1)
a = np.concatenate((a,t))

m = (a[:,:3]==D).all(1)
av = np.concatenate([av]+[av[m]]*2)
ar = np.concatenate([ar]+[ar[m]]*2)
t = a[m]
t[:,5:] = t[:,:4:-1]
m = a[m]
m[:,5] = 3 - t[:,5:].sum(1)
a = np.concatenate((a,t,m))

d = a[:,3::2]*2+a[:,4::2]-(a[:,4::2]>a[:,3::2])
av = 1 / av
```

```python
import matplotlib.pyplot as plt
plt.scatter(d[:,0], av)
plt.xlabel('df (01 02 10 12 20 21)');
```

```python
plt.scatter(d[:,1], av)
plt.xlabel('d')
```

```python
plt.figure(figsize=(18,12))
N = np.arange(len(a),dtype='H')
reg = {'min':((D/np.median(a[:,:3],1)+1)*(D/a[:,:3].max(1))+1)*4, 'theory':((D/a[N,a[:,6]]+1)*(D/a[N,3-a[:,5:].sum(1)])+1)*4, 'actual':ar}
for i,v in enumerate(reg):
    for j,w in enumerate(reg):
        plt.subplot(len(reg),len(reg),i*len(reg)+j+1).scatter(reg[v] if i==j else reg[v]/reg[w], av)
        plt.xscale('log')
        plt.xlabel(v if i==j else v+'/'+w)
```
